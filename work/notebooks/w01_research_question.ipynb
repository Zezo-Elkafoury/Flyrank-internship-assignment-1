## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/zezo-Elkafoury/Flyrank-internship-assignment-1/main/data/raw/content_refresh_anonymized.csv')
df.head()
```

```python
summary = pd.DataFrame({
    "Metric": [
        "Rows",
        "Columns",
        "Missing Values",
        "Duplicate Rows"
    ],
    "Value": [
        len(df),
        len(df.columns),
        df.isna().sum().sum(),
        df.duplicated().sum()
    ]
})

summary
```

```python
# Average CTR for each trend direction
df.groupby('trend_direction')['ctr'].mean().plot(kind='bar')
```

*This shows that flat pages get lots of clicks from users even though they are not growing, which might indicate that they are good candidates for review and refresh*

```python
# Are there pages in page_1 of search engine that are declining
df[df['position_tier'] == 'page_1']['trend_direction'].value_counts()
```

*It seems interesting to find out that the majority of page_1 pages are going down which also shows that they may be good candidates for refresh*
